```python
def format_dollars(value):
    if value >= 1_000_000:
        return f"{value / 1_000_000:.2f}M"
    elif value >= 1_000:
        return f"{value / 1_000:.2f}K"
    else:
        return str(value)
```

## What is the 95% confidence intervals for profit margins of top-performing and low-performing vendors

```python
top_threshold = df["TotalSalesDollars"].quantile(0.75)
low_threshold = df["TotalSalesDollars"].quantile(0.25)
```

```python
top_vendors = df[df["TotalSalesDollars"] >= top_threshold]["ProfitMargin"].dropna()
low_vendors = df[df["TotalSalesDollars"] <= low_threshold]["ProfitMargin"].dropna()

# Remove extreme outliers using IQR
def remove_outliers(series):
    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    return series[(series >= lower_bound) & (series <= upper_bound)]

top_vendors = remove_outliers(top_vendors)
low_vendors = remove_outliers(low_vendors)
```

```python
top_vendors
```

```python
def confidence_interval(data, confidence=0.95):
    mean_val = np.mean(data)
    std_err = np.std(data, ddof=1) / np.sqrt(len(data))  # Standard error
    t_critical = stats.t.ppf((1 + confidence) / 2, df=len(data) - 1)
    margin_of_error = t_critical * std_err

    return mean_val, mean_val - margin_of_error, mean_val + margin_of_error
```

```python
top_mean, top_lower, top_upper = confidence_interval(top_vendors)
low_mean, low_lower, low_upper = confidence_interval(low_vendors)

print(f"Top Vendors 95% CI: ({top_lower:.2f}, {top_upper:.2f}), Mean: {top_mean:.2f}")
print(f"Low Vendors 95% CI: ({low_lower:.2f}, {low_upper:.2f}), Mean: {low_mean:.2f}")

plt.figure(figsize=(12, 6))

# Top Vendors Plot
sns.histplot(top_vendors, kde=True, color="blue", bins=30, alpha=0.5, label="Top Vendors")
plt.axvline(top_lower, color="blue", linestyle="--", label=f"Top Lower: {top_lower:.2f}")
plt.axvline(top_upper, color="blue", linestyle="--", label=f"Top Upper: {top_upper:.2f}")
plt.axvline(top_mean, color="blue", linestyle="-", label=f"Top Mean: {top_mean:.2f}")

# Low Vendors Plot
sns.histplot(low_vendors, kde=True, color="red", bins=30, alpha=0.5, label="Low Vendors")
plt.axvline(low_lower, color="red", linestyle="--", label=f"Low Lower: {low_lower:.2f}")
plt.axvline(low_upper, color="red", linestyle="--", label=f"Low Upper: {low_upper:.2f}")
plt.axvline(low_mean, color="red", linestyle="-", label=f"Low Mean: {low_mean:.2f}")
# Finalize Plot
plt.title("Confidence Interval Comparison: Top vs. Low Vendors (Profit Margin)")
plt.xlabel("Profit Margin (%)")
plt.ylabel("Frequency")
plt.legend()
plt.grid(True)
plt.xlim(left=0)
plt.show()
```

## Insights: Profit Margin Comparison — Top vs. Low-Performing Vendors

**Confidence Interval Summary (95% CI)**
- **Top-performing vendors** (top 25% by `TotalSalesDollars`): CI = 30.49% – 31.16%, Mean = 30.82%
- **Low-performing vendors** (bottom 25% by `TotalSalesDollars`): CI = 8.50% – 13.26%, Mean = 10.88%

**Key Observations**
- Top vendors maintain a notably higher and tighter profit margin range, while low vendors show a wider, more dispersed margin distribution — including a long tail extending toward 100%.
- The narrow CI for top vendors suggests their margins are consistent and predictable across the group.
- The wide CI for low vendors indicates high variability — some low-sales vendors operate at very low margins, while a smaller subset shows unusually high margins, which may warrant a separate data quality check.

**Possible Explanations (to validate further)**
- Higher-sales vendors may benefit from economies of scale, stronger supplier bargaining power, or more efficient operations.
- Lower-sales vendors may face higher relative costs, less pricing leverage, or inconsistent demand — though this dataset doesn't directly confirm the cause.

**Recommendations**
- **For low-performing vendors:** Investigate cost structure, renegotiate supplier terms, or review pricing strategy on low-margin items.
- **For top-performing vendors:** Maintain current efficiency while monitoring competitive pricing pressure as sales scale.
- **Next step:** Examine the low-vendor outliers near 100% margin to confirm they reflect real business cases rather than data artifacts (e.g., very small `TotalSalesDollars` denominators).

*Note: This analysis is based on a sales-volume split (top/bottom 25%), not an independent performance score — "low-performing" here specifically means low sales volume, not poor business quality.*
